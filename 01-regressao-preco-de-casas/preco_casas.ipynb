# Projeto: Prevendo o Preço de Casas com Machine Learning

Este notebook tem como objetivo ensinar de forma prática como criar um modelo de machine learning, usando python, pandas, matplotlib e Scikit-learn.

Você vai aprender:
- Como importar dados de um CSV;
- Como visualizar os dados;
- Como treinar um modelo de Regreção Linear;
- Como fazer previsões com base em novos dados.

```python
# PASSSO 1: Fazer upload do arquivo csv

from google.colab import files
import pandas as pd

uploaded = files.upload()

# Lendo o CSV
df = pd.read_csv(next(iter(uploaded)), sep = ',')
df.head(n=14)
```

```python
# PASSO 2: Analisar os dados

# Exibir informações gerais sobre os dados

df.info()

# Exibir estatísticas básicas
df.describe
```

```python
# PASSO 3: Visualizar a relação entre área e preço

import matplotlib.pyplot as plt

plt.scatter(df['Área (m²)'],df['Preço (R$)'])
plt.xlabel('Área (m²)')
plt.ylabel('Preço (R$)')
plt.title('Área X Preço das Casas')
plt.show()
```

```python
# PASSO 4: Treinar um modelo de Regressão Linear

from sklearn.linear_model import LinearRegression
import numpy as np

# Separando variáveis de entrada (x) e saída (y)

x = df[['Área (m²)', 'Quartos']].values
y = df['Preço (R$)'].values

# Criando o modelo

modelo = LinearRegression()

# Treinando o modelo com os dados

modelo.fit(x,y)

print("Modelo treinado com sucesso!")
```

```python
# PASSO 5: Fazendo previsões

# Exemplo: prever o preço de uma casa com  85m² e 2 quartos.

nova_casa = np.array([[85,2]])
previsao = modelo.predict(nova_casa)

print(f"Preço previsto para uma casa com 85m² e 2 quartos: R$ {previsao[0]:.2f} ")
```

```python
# PASSO 6: Teste com outros valores

# Altere os valores abaixo e veja as previsões do modelo

area = float(input("Digite a área da casa: "))
quartos = int(input("Digite a quantidade de quartos: "))

nova_casa = np.array([[area,quartos]])

previsao2 = modelo.predict(nova_casa)

print(f"Preço previsto: R$ {previsao2[0]:.2f}")
```
