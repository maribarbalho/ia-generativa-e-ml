# Projeto: Prevendo o Consumo de Combustível com Machine Learning

Este notebook tem como objetivo ensinar de forma prática como criar um modelo de machine learning, usando python, pandas, matplotlib e Scikit-learn.

Você vai aprender:
- Como importar dados de um CSV;
- Como visualizar os dados;
- Como treinar um modelo de Regreção Linear;
- Como fazer previsões com base em novos dados.

```python
# PASSSO 1: Fazer upload do arquivo csv

from google.colab import files
import pandas as pd

uploaded = files.upload()

# Lendo o CSV
df = pd.read_csv(next(iter(uploaded)), sep = ',')
df.head(n=14)
```

```python
# PASSO 2: Analisar os dados

# Exibir informações gerais sobre os dados

df.info()

# Exibir estatísticas básicas
df.describe
```

```python
# PASSO 3

import matplotlib.pyplot as plt

plt.scatter(df['Peso (kg)'], df['Consumo (km/L)'])
plt.xlabel('Peso do carro (kg)')
plt.ylabel('Consumo (km/L)')
plt.title('Peso do Carro X Consumo de Combustível')
plt.grid(True)
plt.show()
```

```python
# PASSO 4: Treinar um modelo de Regressão Linear

from sklearn.linear_model import LinearRegression
import numpy as np

# Separando variáveis de entrada (x) e saída (y)
x = df[['Peso (kg)', 'Potência (HP)', 'Cilindros']].values
y = df['Consumo (km/L)'].values

# Criando o modelo
modelo = LinearRegression()

# Treinando o modelo com os dados
modelo.fit(x, y)

print("Modelo treinado com sucesso!")
```

```python
# PASSO 5: Fazendo previsões

# Exemplo: prever o consumo de um carro com 1500kg, 110 HP e 4 cilindros

nova_amostra = np.array([[1500, 110, 4]])
previsao = modelo.predict(nova_amostra)

print(f"Consumo previsto: {previsao[0]:.2f} km/L")
```

```python
# PASSO 6: Teste com outros valores

# Altere os valores abaixo e veja as previsões do modelo

peso = float(input("Digite o peso do carro (kg): "))
potencia = float(input("Digite a potência do motor (HP): "))
cilindros = int(input("Digite o número de cilindros: "))

nova_amostra = np.array([[peso, potencia, cilindros]])

previsao2 = modelo.predict(nova_amostra)

print(f"Consumo previsto: {previsao2[0]:.2f} km/L")
```
