# Projeto: Geração de imagens com IA Generativa no Google Colab (Stable Diffusion + Hugging Face)

## Objetivo:
Neste projeto, vamos usar a IA Generativa de Imagens (Stable Diffusion) através da biblioteca Diffusers da Hugging Face para criar imagens a partir de textos (prompts).

```python
# Instalar as bibliotecas necessárias
# difusers: Baixar e usar o modelo de IA de Geração de Imagem
# transformers: Entender e processar o texto do prompt
# accelerate: Otimizar e gerenciar o uso de CPU/GPU
# scipy: Fazer cálculos matemáticos necessários internamente
# safetensons: Carregar o modelo de IA de forma rápida e segura

!pip install diffusers transformers accelerate scipy safetensors
```

```python
# Importar as bibliotecas

from diffusers import StableDiffusionPipeline
import torch
```

```python
# Definir o token de autenticção Hugging Face
# Para criar a conta gratuita: https://huggingface.co/join
# Depois, gerar um token de acesso: https://huggingface.co/settings/tokens

from getpass import getpass

token = getpass("Cole aqui o seu token do Hugging Face: ")
```

```python
# Carregar o modelo Stable Diffusion da Hugging Face

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
    use_auth_token=token
).to("cuda")
```

```python
# Gerar uma imagem a partir de um prompt de texto

prompt = "a piece of cake with a flower"

image = pipe(prompt).images[0]

# Exibir a imagem gerada
image.show()

# Salvar a imagem localmente
image.save("imagem_gerada.png")
```

```python
# Agora é sua vez: Experimente com outros prompts!

prompt = input("Digite um novo prompt para gerar uma imagem:")
image = pipe(prompt).images[0]
image.show()

# Salvar a imagem localmente

image.save("imagem_gerada2.png")
```
