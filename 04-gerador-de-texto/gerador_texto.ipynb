Projeto prático de geração de texto em português usando a IA Generativa com o Hugging Face Transformers, focado no modelo GPT-2 adaptado para o português.

```python
# Instalar a biblioteca Transformers que contém vários modelos de IA para NLP
# Precisa ser executado uma vez por sessão no Colab.

!pip install transformers
```

Importando as bibliotecas:

```python
# Pipeline (fluxo de como algo deve funcionar): Interface fácil de usar da Hugging FAce
# AutoTokenizer: Carrega o tokenizador do modelo.
# AutoModeForCausalLm: Carrega o modelo da linguagem (ex: GPT-2)

from transformers import pipeline, AutoTokenizer, AutoModelForCausalLM
```

Carregar o modelo em português

```python
# Aqui é feito o download de um GPT-2 pequeno treinado em português
# O modelo e tokenizador são carregados na memória.

modelo_pt = "pierreguillou/gpt2-small-portuguese"
tokenizer = AutoTokenizer.from_pretrained(modelo_pt)
modelo = AutoModelForCausalLM.from_pretrained(modelo_pt)
```

Criar a pipeline de geração de texto.

```python
# Essa pipeline recebe um texto inicial (prompt) e gera uma continuação

gerador = pipeline("text-generation", model=modelo, tokenizer=tokenizer)
```

Entrada do usuário

```python
# O usuário digita uma frase inicial para o modelo continuar

prompt = input("Digite o começo do texto (ex: Era uma vez...) ")
```

Geração do texto

```python
# O modelo gera uma resposta com no máximo 100 tokens

resposta = gerador(prompt, max_length = 100, num_return_sequences=1)[0]["generated_text"]
```

Exibição do resultado

```python
# Exibe o texto gerado para o usuário

print("\n Texto gerado em português: ")
print(resposta)
```

```python
# Baixando a biblioteca Docx

!pip install python-docx

# Importando a biblioteca Docx para manipulação de documentos

from docx import Document

# Criando documento word com a história

doc = Document()
doc.add_heading("História gerada por IA", 0)
doc.add_paragraph(resposta)

# Salvando o documento
doc.save("historia_gerada.docx")

print("n\ Documento 'historia_gerada.docx' salvo com sucesso!")
```
